## Object Oriented Programming

Understanding the 4 principles of OOP.
1. Inheritance: Passing on properties and behaviors to child class
2. Abstraction: Hiding implementaion details
3. Encapsulation: Shielding properties
4. Polymorphism: Able to exist in many forms

The phrases above are carefully choses words to drive home the understanding of these concepts

### Inheritance: Simply means parent passing onto child it's properties and behaviors
We will demonstrate with Person class as parent and Student class as child.

```python
# A parent class with properties first_name and age
class Person:
    def __init__(self, first_name, age):
        self.__first_name = first_name
        self.age = age

    def get_first_name(self):
        return self.__first_name
    

# A child class created using parent.
class Student(Person):
    def __init__(self, first_name, age, student_id):
        super().__init__(first_name, age)
        self.student_id = student_id

# Use the student class
student = Student('Andrew', 25, 'W05123')
student.get_first_name()
```

It is important to note that the principles states that instead of creating a student class from scratch, use its parent to do so.
1. Relationship is well established. The two classes working together
2. Repitition is reduced (DRY concept). Volume of code is reduced
3. One central control - A design modification of the properties will only be done at the parent class. `change first_name to enforce upper case`

### Encapsulation: Shielding property 

It is important to note that encapsulation is abot the property and not the behavior. Also, be aware that actions on the property is either to get it or to set it.

1. get it -> student.first_name -> 'Andrew' 
2. set it -> student.first_name = 'Jacob'


```python
student.__first_name = 'Jacob' # This has exposed this attribute to change. It is not shielded
student.get_first_name()
```

### Abstraction: Hiding implementation details
1. connect
2. authenticate
3. send email
4. disconnect

```python

class SendMail:
    def __init__(self):
        pass

    def __connect(self):
        print("Server connected")

    def send(self):
        self.__connect()
        self.__authenticate()
        print("Email is sent")
        self.__disconnect()

    def __disconnect(self):
            print("Server is disconnect")

    def __authenticate(self):
        print("sender is authenticated")

email = SendMail()
email.send()
    
```

### Polymorphism: Many forms
We can build a system that will implement unique behavior based on request

```python
import math
from abc import ABC, abstractmethod

class Shape(ABC):
    
    def __init__(self, size):
        self.size = size

    @abstractmethod
    def area(self):
        pass


class Circle(Shape):
    def __init__(self, size):
        super().__init__(size)

    def area(self):
        return f'Area of circle: {math.pi * self.size**2}'

class Square(Shape):
    def __init__(self, size):
        super().__init__(size)

    def area(self):
        return f'Area of square: {self.size**2}'



# defining a function that will polymorphic 
def getArea(shape: Shape):
    return shape.area()

print(getArea(Square(4)))
```
