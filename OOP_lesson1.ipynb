### Define a class
1. A blueprint for making objects
2. A template for making objects
3. Object factory

```python
class Point:
    pass

# a class is defined with no characteristics
point = Point()
point
```

### A class variable x

```python
class Point:
    x = 5

x = 10 # not a property of the class

print(x, point.x) # x is a property of the class
```

### class method

```python
# a class with two class variables x & y and one class method add()
class Point:
    x = 5
    y = 7

    def add():
        return Point.x + Point.y
    
print(Point.add())

# a class with no class variable 
class Point:
    def add(x, y):
        return x + y
    
print(Point.add(2, 3))
```

### What is an object?
1. An instance of the class
2. A sample of the class
3. A by-product of the class

Let me take you back to the previous class with add method.

```python
class Point:
    def add(self):
        return 2 + 3
    
point = Point()
print(point.add())

# Below is the complete form of the class above
class Point:
    def __init__(self): # belong to the object
        pass

    def add(self): # belong to the object
        return 2 + 3
    
print(Point.add()) # error: add is an object method (instance method)
```
