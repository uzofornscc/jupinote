## Adapter
There are viable reasons why we may have to leave classes the way they are or exist.
1. It is being consumed already
2. You do not have access to modify it
Sometimes classes have been written and you don't have the option of modifying their interface to
 suit your needs. This happens if the method you are calling is on a different system across a
 network, a library that you may import or generally something that is not viable to modify directly for
 your particular needs.

```python
# This is a setup such that a user have one and therefore cannot have the other, you are tasked to 
# help the user benefit from the feature of the system which she does not have

# a class that do not need to be modified
class Canadian:

    def voltage(self):
        return 120

# another independent class 
class European:

    def voltage(self):
        return 230

# a class built to adapt or conform to either class above  
class Adapter:
    def __init__(self, source):
        match source:
            case Canadian():
                self.source = European()
            case European():
                self.source = Canadian()
            case _:
                self.source = None
        
    def voltage(self):
        if self.source: 
            return self.source.voltage()
                  
source = European()

adapter = Adapter(source)
print(adapter.voltage())



```

### A system exist such that there is need for upgrade. The upgrade is to ensure the already existing system will still work as expected without modification. 
An Audio player (plays only mp3) already exist and there is need to enhance the system to be able to play video (.mp4 and .vlc file format) and yet later upgrade for .wav file format.

```python

from pathlib import Path

class AudioPlayer:

    def play(self, file):
        print(f"Playing mp3 !!! - {file}")

class MP4_Player:

    def play(self, file):
        print(f"Playing mp4 file !!! - {file}")

class Vlc_Player:

    def play(self, file):
        print(f"Playing vlc file !!! - {file}")

class WaveMedia:

    def play(self, file):
        print(f"playing wave - {file}")


class MediaAdapter:

    def __init__(self, player):
        self.player = player

    def _get_file_type(self, file):
        ext = Path(file).suffix.lower().lstrip('.')
        player = self.player.get(ext)
        return player
            
    def play(self, file):
        player = self._get_file_type(file)
        if player:
            player.play(file)
        else:
            print("Unacceptable file format")
                    
# model                
player = {
            'mp4': MP4_Player(),
            'vlc': Vlc_Player(),
            'mp3': AudioPlayer(),
            'wav': WaveMedia()
        }
media = MediaAdapter(player)
media.play('file.wav')
```
